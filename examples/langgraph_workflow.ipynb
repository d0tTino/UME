# LangGraph Workflow

Demonstrate using the `LangGraph` adapter in a realistic workflow.

```
from ume.integrations.langgraph import LangGraph
client = LangGraph()
```

```
# Create a node
client.send_events([{"event_type": "CREATE_NODE", "timestamp": 1, "node_id": "node1", "payload": {"node_id": "node1", "attributes": {"text": "hello"}}}])
```

```
# Recall a node
print(client.recall({"node_id": "node1"}))
```
