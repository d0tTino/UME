# Letta Workflow

Demonstrate using the `Letta` adapter in a realistic workflow.

```
from ume.integrations.letta import Letta
client = Letta()
```

```
# Create a node
client.send_events([{"event_type": "CREATE_NODE", "timestamp": 1, "node_id": "node1", "payload": {"node_id": "node1", "attributes": {"text": "hello"}}}])
```

```
# Recall a node
print(client.recall({"node_id": "node1"}))
```
