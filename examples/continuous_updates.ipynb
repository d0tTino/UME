# Continuous Memory Updates

```python
import asyncio
from ume.integrations.langgraph import AsyncLangGraph

client = AsyncLangGraph()
```

```python
async def main():
    async with client:
        for i in range(3):
            await client.send_events([{"event_type": "CREATE_NODE", "timestamp": i, "payload": {"node_id": f'n{i}', "attributes": {}}}])
            async for node in client.path_stream({"source": 'n0', "target": f'n{i}' }):
                print(node)
            await asyncio.sleep(1)

asyncio.run(main())
```
